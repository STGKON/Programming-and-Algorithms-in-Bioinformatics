```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# https://stackoverflow.com/questions/58633026/matplotlib-and-the-factorial-function
# The gamma function is a generalization of the factorial function to real and complex numbers
# For non-negative integer n: factorial(n) = gamma(n+1)
from scipy.special import gamma
```

```python
# A simple implementation of the factorial function
def fact(n):
  if n <= 1:
    return 1
  else:
    return n * fact(n-1)
```

```python
def nlogn(n):
  return n * np.log2(n)
```

```python
# Create 1000 evenly spaced numbers and store them in x
x = np.linspace(0.59, 1.18, 1000)

plt.plot(x, x, label='x')
plt.plot(x, nlogn(x), label='x log(x)')
plt.plot(x, x**5, label='$x^5$')
plt.plot(x, 2**x, label='$2^x$')
plt.plot(x,gamma(x), label='Factorial x!')

plt.ylim(-4,4)
plt.legend()
```
